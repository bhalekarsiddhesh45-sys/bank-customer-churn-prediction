# 04 Feature engineering

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("C:/Users/Siddhesh/Downloads/bank-customer-churn/data/processed/Bank_churn_prediction.csv")

df.head()
```

```python
df["HasZeroBalance"] = (df["Balance"] == 0).astype(int)
df[["Balance", "HasZeroBalance"]].head()
```

```python
df["BalanceToSalary"] = (
    df["Balance"] / df["EstimatedSalary"].replace(0, np.nan)
)

df[["Balance", "EstimatedSalary", "BalanceToSalary"]].head()
```

```python
df[["Age", "AgeBand"]].head()
```

```python
df["ProductsGroup"] = df["NumOfProducts"].clip(upper=3)

df[["NumOfProducts", "ProductsGroup"]].head()
```

```python
df["TenureYoungAccount"] = (df["Tenure"] <= 1).astype(int)

df[["Tenure", "TenureYoungAccount"]].head()
```

```python
new_features = [
    "HasZeroBalance",
    "BalanceToSalary",
    "AgeBand",
    "ProductsGroup",
    "TenureYoungAccount"
]

df[new_features].head()
```
